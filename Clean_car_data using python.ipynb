## fixing incosistent formating

```python
df.dtypes
```

## 1.resale_price column.

```python
# we remove rupeeya symbol and convert lakh or crore into hundred.
x=df["resale_price"]
def convert_amount(x):
    if pd.isna(x):
        return np.nan

    x = str(x).strip()

    x = x.replace("₹","").replace(",","")

    if "Lakh" in x:
        return float(x.replace("Lakh","").strip()) * 100000

    elif "Crore" in x:
        return float(x.replace("Crore","").strip()) * 10000000

    else:
        return np.nan
df["resale_price"]=df["resale_price"].apply(convert_amount)
```

```python
df.head(1)
```

## 2.full_name column.

```python
# extract year from the full_name column.
x=df["full_name"]
def ext_year(x):
    y=x.split(" ")
    return y[0]
df["registered_year"]=df["full_name"].apply(ext_year)
```

```python
# extract car_name from the full_name column
x=df["full_name"]
def ext_car_name(x):
   y=x.split(" ")
   return " ".join(y[2:5])
df["car_name"]=df["full_name"].apply(ext_car_name)
```

```python
df["car_name"].head(10)
```

```python
# extract company_name from the full_name column.
x=df["full_name"]
def company_name(x):
      y=x.split(" ")
      return y[1]

df["full_name"]=df["full_name"].apply(company_name)
df.rename(columns={"full_name": "company_name"}, inplace=True)
```

## 3. kms_driven column.

```python
x=df["kms_driven"]
def ext_numbers(x):
    return x.replace(" Kms","").replace(",","").strip()
df["kms_driven"]=df["kms_driven"].apply(ext_numbers)
```

```python
df["kms_driven"].head(10)
```

## 4.Mileage column

```python
x=df["mileage"]
def remove_kmpl(x):
     return x.replace("kmpl","").replace("km/kg","").strip()
df["mileage"]=df["mileage"].apply(remove_kmpl)
```

## 5.engine_capacity column.

```python
# we remove "0 cc" in the engine_capactiy column.because 0cc engine capacity is not available.
df = df[df["engine_capacity"].astype(str) != "0 cc"]
```

```python
# we remove "Not available" values from the engine_capacity.
df=df[df["engine_capacity"].astype(str) != "Not available"]
```

```python
# we remove "cc" from the engine_column.
df.loc[:, "engine_capacity"] = (
    df["engine_capacity"]
    .str.replace("cc", "", regex=False)
    .str.strip()
)
```

```python
df["engine_capacity"].head(3)
```

## 6.Insurance column

```python
df=df[df["insurance"].astype(str) !="1"]
df=df[df["insurance"].astype(str) != "2"]
df=df[df["insurance"].astype(str) != "Not Available"]
```

```python
x=df["insurance"]
def change_third_p(x):
    return x.replace("Third Party insurance","Third Party").strip()
df["insurance"]=df["insurance"].apply(change_third_p)
```

```python
df["insurance"].head(10)
```

## 7.owner type column

```python
df=df[df["owner_type"].astype(str) != "Not available"]
```

## Changing Datatype

```python
# kms_driven column datatype "object" change into "int64"
df["kms_driven"]=df["kms_driven"].astype("int64")
```

```python
# registered_year column datatype is "object" change into "int64".
df["registered_year"]=df["registered_year"].astype("int64")
```

```python
# Seats column datatype is "floats" change into "int64".
df["seats"]=df["seats"].astype("int64")
```

```python
# mileage column datatype is "object" changed into "int64".
df["mileage"]=df["mileage"].astype(float)
```

```python
# engine_capacity column datatype is "object"  changed into "int64'.
df["engine_capacity"]=df["engine_capacity"].astype("int64")
```

```python
df.dtypes
```

```python
df["resale_price"]=df["resale_price"].fillna(df["resale_price"].median())
```
